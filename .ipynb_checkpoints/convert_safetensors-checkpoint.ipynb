```python
# convert_safetensors_to_bin.py
from safetensors.torch import load_file
import torch

# paths—adjust to your actual files
in_path  = "./output/Med3DVLM-Qwen-2.5-7B/mm_projector.safetensors"
out_path = "./output/Med3DVLM-Qwen-2.5-7B/mm_projector.bin"

# load the safetensors weights into a state dict
state_dict = load_file(in_path, device="cpu")

# save as a PyTorch pickle
torch.save(state_dict, out_path)
print(f"Converted {in_path} → {out_path}")
```
